```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
from sklearn.preprocessing import LabelEncoder
import joblib
```

```python
filename= "ObesityDataSet_raw_and_data_sinthetic.csv"
df = pd.read_csv(filename)
df.head()
```

```python
df.info()
```

```python
# Checking nulls
print(df.isnull().sum())
df_encoded = df.copy()

# Encode categorical variables
label_encoders = {}


for col in df.select_dtypes(include='object').columns:
    le = LabelEncoder()
    df_encoded[col] = le.fit_transform(df[col])  # Apply the encoding
    label_encoders[col] = le  # Store the encoder

df_encoded.head()
```

```python
# Save the label encoders to a file
joblib.dump(label_encoders, "label_encoders.joblib")
```

```python
#Target Variable Distribution

plt.figure(figsize=(10, 5))
sns.countplot(x='NObeyesdad', data=df)
plt.xticks(rotation = 45)
plt.title("Distribution of Obesity Levels")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(x='NObeyesdad', hue='Gender', data=df)
plt.xticks(rotation=45)
plt.title("Obesity Level by Gender")
plt.show()
```

```python
df_num = df.select_dtypes(include='number')
plt.figure(figsize=(10, 6))
sns.heatmap(df_num.corr(), annot=True, cmap='coolwarm')
plt.title("Feature Correlation Matrix")
plt.show()
```

```python
sns.pairplot(df[['Age', 'Height', 'Weight', 'FAF', 'TUE', 'NObeyesdad']], hue='NObeyesdad', diag_kind='kde')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.boxplot(x='NObeyesdad', y='FAF', data=df)
plt.xticks(rotation=45)
plt.title("Physical Activity (FAF) by Obesity Level")
plt.show()
```

```python
caec_counts = pd.crosstab(df['CAEC'], df['NObeyesdad'])
caec_counts.plot(kind='bar', stacked=True, figsize=(10, 6), colormap='viridis')
plt.title("Eating Between Meals vs Obesity Level")
plt.ylabel("Number of People")
plt.xticks(rotation=0)
plt.show()
```

```python
sns.displot(df, x="Age", hue="NObeyesdad", kind="kde", height=5, aspect=2)
plt.title("Age Density across Obesity Levels")
plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

features = ['Weight', 'Height', 'Age', 'FAF', 'TUE', 'CH2O']
X_scaled = StandardScaler().fit_transform(df[features])
kmeans = KMeans(n_clusters=4)
df['Cluster'] = kmeans.fit_predict(X_scaled)

sns.pairplot(df, vars=features, hue='Cluster')
plt.show()
```
